# Quantum-Greedy-MVC: Basic Usage\n\nThis notebook shows the simplest way to solve MVC and MIS with the package API.

```python
import networkx as nx\nfrom quantum_greedy_mvc import QuantumGreedySolver
```

```python
G = nx.cycle_graph(6)\nweights = {node: 1.0 for node in G.nodes()}
```

```python
solver = QuantumGreedySolver(method="qeg_ldf", qeg_time=0.35, qeg_trotter_layers=1)\nmvc = solver.solve_mvc(G, weights=weights)\nmis = solver.solve_mis(G, weights=weights)\n\nprint('MVC solution:', sorted(mvc.solution), 'objective:', mvc.objective)\nprint('MIS solution:', sorted(mis.solution), 'objective:', mis.objective)
```
